## 1. O que é uma Função?
Uma função é um bloco de código autocontido que só executa quando é chamado.
* **Sintaxe:**
```python
def nome_da_funcao(parametro1, parametro2):
    # Processamento
    return resultado
```


````python
# Exemplo do Professor: Função para calcular área de um retângulo
def calcular_area_terreno(largura, comprimento):
    area = largura * comprimento
    return area

# Chamando a função
terreno1 = calcular_area_terreno(10.0, 25.0)
terreno2 = calcular_area_terreno(8.0, 20.0)

print(f"Área do terreno 1: {terreno1} m²")
print(f"Área do terreno 2: {terreno2} m²")
````

## 2. Diferença Crucial: `return` vs `print()`
* `print()` apenas exibe uma mensagem visual no terminal (não guarda o resultado para cálculos posteriores).
* `return` devolve o dado processado para quem chamou a função, permitindo armazená-lo em uma variável.


````python
# Demonstrando a diferença
def soma_com_print(a, b):
    print("Resultado no print:", a + b)

def soma_com_return(a, b):
    return a + b

ret1 = soma_com_print(4, 6)
ret2 = soma_com_return(4, 6)

print("Valor armazenado em ret1:", ret1)  # None!
print("Valor armazenado em ret2:", ret2)  # 10!
````
