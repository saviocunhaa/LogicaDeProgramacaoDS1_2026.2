## 1. Escopo de Variáveis (Local vs Global)
* **Variável Local:** Criada dentro de uma função. Só existe durante a execução daquela função.
* **Variável Global:** Criada fora de qualquer função. Pode ser lida em todo o arquivo.


```python
# Exemplo de Escopo
escola = "EEEP Prof. Sebastião Vasconcelos Sobrinho" # Global

def apresentar_turma():
    turma = "1º Ano DS" # Local
    print(f"Turma: {turma} da instituição {escola}")

apresentar_turma()
# print(turma) # Descomentar essa linha causa erro NameError!
```

## 2. Funções Anônimas (`lambda`)
São funções curtas de apenas uma linha:
`lambda argumentos: expressao`


```python
# Exemplo de lambda para converter Celsius em Fahrenheit
celsius_para_fahrenheit = lambda c: (c * 9/5) + 32

print("30°C em Fahrenheit:", celsius_para_fahrenheit(30))
```
