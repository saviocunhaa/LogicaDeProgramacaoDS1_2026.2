## 1. Operadores Relacionais (Comparação)

| Operador | Significado | Exemplo | Resultado |
| :---: | :--- | :--- | :---: |
| `==` | Igual a | `5 == 5` | `True` |
| `!=` | Diferente de | `5 != 3` | `True` |
| `>` | Maior que | `10 > 20` | `False` |
| `<` | Menor que | `10 < 20` | `True` |
| `>=` | Maior ou igual a | `7 >= 7` | `True` |
| `<=` | Menor ou igual a | `4 <= 3` | `False` |


```python
# Exemplos de comparações
idade = 17
tem_titulo = True

print("Maior de idade?", idade >= 18)
print("Idade diferente de 18?", idade != 18)
```

## 2. Operadores Lógicos

* **`and` (E):** Retorna `True` **apenas** se todas as condições forem verdadeiras.
* **`or` (OU):** Retorna `True` se **pelo menos uma** condição for verdadeira.
* **`not` (NÃO):** Inverte o valor lógico (`True` vira `False` e vice-versa).

### Tabela-Verdade Resumida:
* `True and True` ➔ `True`
* `True and False` ➔ `False`
* `False or True` ➔ `True`
* `False or False` ➔ `False`
* `not True` ➔ `False`


```python
# Exemplo de verificação de permissão
aluno_matriculado = True
mensalidade_em_dia = True
tem_carteirinha = False

pode_entrar_no_lab = aluno_matriculado and (mensalidade_em_dia or tem_carteirinha)
print("Acesso autorizado ao laboratório?", pode_entrar_no_lab)
```
