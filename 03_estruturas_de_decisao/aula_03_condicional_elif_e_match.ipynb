## 1. Estrutura `if - elif - else`
Quando temos mais de duas possibilidades exclusivas:

```python
if condicao_1:
    bloco_1
elif condicao_2:
    bloco_2
elif condicao_3:
    bloco_3
else:
    bloco_padrao
```


````python
# Exemplo do Professor: Classificação de notas
nota = float(input("Digite a média final do estudante: "))

if nota >= 9.0:
    conceito = "Excelente (A)"
elif nota >= 7.0:
    conceito = "Bom (B)"
elif nota >= 6.0:
    conceito = "Regular (C)"
else:
    conceito = "Reprovado / Recuperação (D)"

print(f"Conceito final: {conceito}")
````

## 2. Estrutura `match - case`
Ideal para menus de opções ou seleção baseada em valores fixos:

```python
opcao = int(input("Escolha: 1-Pix, 2-Cartão, 3-Boleto: "))

match opcao:
    case 1:
        print("Pagamento instantâneo via Pix.")
    case 2:
        print("Pagamento no Cartão de Crédito.")
    case 3:
        print("Boleto bancário gerado.")
    case _:
        print("Opção inválida!")
```
