## 1. Modos de Abertura de Arquivos
* `'w'` (Write): Cria o arquivo para escrita. Se o arquivo já existir, **sobrescreve todo o conteúdo**.
* `'a'` (Append): Cria o arquivo ou **anexa novos dados ao final**, preservando o histórico existente.
* `'r'` (Read): Abre o arquivo em modo apenas leitura. Se não existir, gera erro.


```python
# Exemplo de Escrita com 'w'
with open("exemplo_turma.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Curso Técnico em Desenvolvimento de Sistemas\n")
    arquivo.write("EEEP Prof. Sebastião Vasconcelos Sobrinho\n")

print("Arquivo gerado com sucesso!")
```

## 2. Lendo o Arquivo com `'r'` e Tratando Exceções
Sempre utilize `try / except` para evitar que seu programa quebre caso o arquivo não seja encontrado.


```python
# Exemplo de Leitura Segura
try:
    with open("exemplo_turma.txt", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()
        print("Conteúdo lido do arquivo:\n")
        print(conteudo)
except FileNotFoundError:
    print("Aviso: O arquivo solicitado não foi encontrado no disco.")
```
