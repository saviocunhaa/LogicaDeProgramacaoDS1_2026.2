## 1. Dicionários (`dict`)
Dicionários armazenam dados associando uma chave descritiva a um valor:
`dicionario = {"chave": valor}`


```python
# Criando o perfil de um aluno
aluno_ds = {
    "matricula": "2026-DS-014",
    "nome": "Ryan Mesquita",
    "curso": "Desenvolvimento de Sistemas",
    "modulo_atual": 6
}

print("Nome do estudante:", aluno_ds["nome"])
print("Curso:", aluno_ds.get("curso"))

# Adicionando uma nova chave
aluno_ds["status"] = "Ativo"
print("Dicionário atualizado:", aluno_ds)
```

## 2. Conjuntos (`set`)
Conjuntos são coleções desordenadas que **não aceitam elementos duplicados** e suportam operações matemáticas como união (`|`), interseção (`&`) e diferença (`-`).


```python
turma_a = {"Ana", "Carlos", "Beatriz", "Diego"}
turma_b = {"Carlos", "Diego", "Eduardo", "Fernanda"}

intersecao = turma_a & turma_b # Alunos presentes em ambas
print("Alunos nas duas turmas ao mesmo tempo:", intersecao)
```
