## 1. Tuplas (`tuple`): Estruturas Imutáveis
Diferente das listas, as tuplas **não podem ser alteradas** após a criação.


```python
# Coordenadas geográficas de Tianguá (Latitude, Longitude)
coordenadas_tiangua = (-3.7319, -40.9917)
print("Coordenadas:", coordenadas_tiangua)
print("Latitude:", coordenadas_tiangua[0])

# coordenadas_tiangua[0] = 0.0 # Isso geraria um TypeError!
```

## 2. Matrizes Bidimensionais (Listas de Listas)
Podemos representar tabelas com linhas e colunas combinando listas aninhadas:


```python
# Matriz representando o diário de classe: [Nome, Nota 1, Nota 2]
diario = [
    ["Lucas", 8.5, 9.0],
    ["Mariana", 7.0, 6.5],
    ["Pedro", 9.5, 10.0]
]

# Acessando a nota 1 da Mariana (linha 1, coluna 1)
print("Nota 1 da Mariana:", diario[1][1])

# Percorrendo a matriz com for aninhado
print("\nBoletim da Turma:")
for linha in diario:
    nome = linha[0]
    n1 = linha[1]
    n2 = linha[2]
    media = (n1 + n2) / 2
    print(f"Aluno: {nome} | Média: {media:.1f}")
```
