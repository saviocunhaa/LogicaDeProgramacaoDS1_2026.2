## 1. Criando e Acessando Listas
Listas em Python são sequências ordenadas e mutáveis. Cada elemento possui um índice que começa em `0`.


```python
# Criando uma lista de linguagens
linguagens = ["Python", "Java", "C++", "JavaScript"]

print("Primeiro elemento (índice 0):", linguagens[0])
print("Último elemento (índice -1):", linguagens[-1])
print("Tamanho da lista:", len(linguagens))
```

## 2. Métodos Principais de Listas
* `.append(item)`: Adiciona um item ao final da lista.
* `.insert(posicao, item)`: Insere um item em um índice específico.
* `.pop()`: Remove e retorna o último elemento (ou no índice informado).
* `.remove(valor)`: Remove a primeira ocorrência do valor informado.
* `.sort()`: Ordena a lista em ordem crescente/alfabética.


```python
# Manipulando a lista
alunos = ["Ana", "Bruno", "Carlos"]
alunos.append("Daniela")     # Adiciona no fim
alunos.insert(1, "Beatriz")   # Insere na posição 1
print("Lista após inserções:", alunos)

alunos.remove("Bruno")        # Remove por valor
print("Lista após remoção:", alunos)
```
