## 1. A Função `range()`
A função `range()` gera uma sequência de números inteiros:
* `range(fim)`: de 0 até `fim - 1`.
* `range(inicio, fim)`: de `inicio` até `fim - 1`.
* `range(inicio, fim, passo)`: de `inicio` até `fim - 1`, pulando de `passo` em `passo`.


```python
# Exemplo de contagem
print("Contando de 1 a 5:")
for i in range(1, 6):
    print(f"Passo: {i}")

print("\nContagem regressiva:")
for reg in range(5, 0, -1):
    print(reg)
print("Decolagem! 🚀")
```

## 2. Acumuladores e Contadores
Podemos somar valores iterativamente durante a execução do laço:


```python
# Somando os primeiros 5 números inteiros
soma_total = 0
for numero in range(1, 6):
    soma_total += numero  # equivalente a soma_total = soma_total + numero

print("Soma acumulada de 1 até 5:", soma_total)
```
