## 1. O Laço `while`
Diferente do `for`, o `while` é ideal quando **não sabemos de antemão quantas repetições acontecerão**.


```python
# Exemplo do Professor: Validação de entrada
resposta = ""
while resposta.lower() != "sim":
    resposta = input("Você aceita os termos de uso do laboratório? (sim/nao): ")

print("Obrigado! Acesso liberado.")
```

## 2. Comandos `break` e `continue`
* **`break`:** Interrompe e encerra o laço imediatamente.
* **`continue`:** Interrompe apenas a repetição atual e salta para a próxima iteração.


```python
# Exemplo com break e continue
contador = 0
while True:
    contador += 1
    if contador == 3:
        print("Pulando o número 3...")
        continue
    if contador > 5:
        print("Atingiu o limite. Encerrando!")
        break
    print("Contador:", contador)
```
